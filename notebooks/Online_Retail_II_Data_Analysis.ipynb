```python
# Remove exact duplicate rows
df_clean = df.drop_duplicates().copy()

print("Original rows:", len(df))
print("Rows after removing exact duplicates:", len(df_clean))
print("Rows removed:", len(df) - len(df_clean))
```

```python
# Check missing values after duplicate removal
missing = df_clean.isnull().sum()

print("Missing values:")
display(missing)

print("\nMissing percentage:")
display((missing / len(df_clean) * 100).round(2))
```

```python
# Compare transactions with and without Customer ID

missing_customer = df_clean[df_clean['Customer ID'].isna()]
known_customer = df_clean[df_clean['Customer ID'].notna()]

print("Rows with missing Customer ID:", len(missing_customer))
print("Rows with Customer ID:", len(known_customer))

print("\nInvoice types with missing Customer ID:")
print(missing_customer['Invoice'].astype(str).str.startswith('C').value_counts())

print("\nInvoice types with known Customer ID:")
print(known_customer['Invoice'].astype(str).str.startswith('C').value_counts())
```

```python
# Create transaction value
df_clean['TotalAmount'] = df_clean['Quantity'] * df_clean['Price']

# Compare missing vs known Customer IDs
customer_id_summary = df_clean.groupby(
    df_clean['Customer ID'].isna()
)['TotalAmount'].agg(['count', 'sum', 'mean'])

customer_id_summary.index = ['Customer ID Present', 'Customer ID Missing']

display(customer_id_summary)
```

```python
# Investigate rows with missing Description

missing_description = df_clean[df_clean['Description'].isna()]

print("Rows with missing Description:", len(missing_description))
print("Total transaction value:", missing_description['TotalAmount'].sum())

print("\nQuantity summary:")
display(missing_description['Quantity'].describe())

print("\nPrice summary:")
display(missing_description['Price'].describe())

print("\nSample rows:")
display(missing_description.head(20))
```

```python
# Remove rows with missing Description and zero Price
before = len(df_clean)

df_clean = df_clean[
    ~(df_clean['Description'].isna() & (df_clean['Price'] == 0))
].copy()

after = len(df_clean)

print("Rows before:", before)
print("Rows removed:", before - after)
print("Rows after:", after)

print("\nRemaining missing Description values:",
      df_clean['Description'].isna().sum())
```

```python
# Check positive vs negative quantity transactions

positive_qty = (df_clean['Quantity'] > 0).sum()
negative_qty = (df_clean['Quantity'] < 0).sum()

print("Positive quantity rows:", positive_qty)
print("Negative quantity rows:", negative_qty)

print("\nNegative quantity percentage:",
      round(negative_qty / len(df_clean) * 100, 2), "%")

print("\nNegative quantity transaction value:",
      df_clean.loc[df_clean['Quantity'] < 0, 'TotalAmount'].sum())

```

```python
# Check cancellation invoices among negative quantities

negative_rows = df_clean[df_clean['Quantity'] < 0].copy()

negative_cancellations = (
    negative_rows['Invoice']
    .astype(str)
    .str.startswith('C')
    .sum()
)

negative_other = len(negative_rows) - negative_cancellations

print("Negative quantity rows:", len(negative_rows))
print("Negative rows with cancellation invoice (C):", negative_cancellations)
print("Negative rows without C invoice:", negative_other)

print("\nCancellation percentage among negative rows:",
      round(negative_cancellations / len(negative_rows) * 100, 2), "%")
```

```python
# Investigate negative quantities that are NOT cancellation invoices

other_negative = df_clean[
    (df_clean['Quantity'] < 0) &
    (~df_clean['Invoice'].astype(str).str.startswith('C'))
].copy()

print("Rows:", len(other_negative))
print("Total value:", other_negative['TotalAmount'].sum())

print("\nTop descriptions:")
display(other_negative['Description'].value_counts().head(20))

print("\nSample records:")
display(other_negative.head(20))
```

```python
# Check negative-price records

negative_price = df_clean[df_clean['Price'] < 0]

print("Negative price rows:", len(negative_price))
print("Total value:", negative_price['TotalAmount'].sum())

display(negative_price)
```

```python
# Investigate all zero-price records
zero_price = df_clean[df_clean['Price'] == 0]

print("Zero-price rows:", len(zero_price))
print("Zero-price rows with positive quantity:",
      (zero_price['Quantity'] > 0).sum())
print("Zero-price rows with negative quantity:",
      (zero_price['Quantity'] < 0).sum())

print("\nDescriptions of zero-price rows:")
display(zero_price['Description'].value_counts().head(30))
```

```python
# Create sales-analysis dataset
# Exclude:
# 1. Negative-price bad-debt adjustments
# 2. Zero-price records

sales_df = df_clean[
    (df_clean['Price'] > 0) &
    (df_clean['Quantity'] != 0)
].copy()

print("Rows in cleaned dataset:", len(df_clean))
print("Rows in sales dataset:", len(sales_df))
print("Rows excluded:", len(df_clean) - len(sales_df))

print("\nRemaining negative prices:",
      (sales_df['Price'] < 0).sum())

print("Remaining zero prices:",
      (sales_df['Price'] == 0).sum())
```

```python
# Calculate revenue for each transaction
sales_df['Revenue'] = sales_df['Quantity'] * sales_df['Price']

print("Total Revenue: £", round(sales_df['Revenue'].sum(), 2))

print("\nRevenue statistics:")
display(sales_df['Revenue'].describe())
```

```python
# Create time-based analysis fields

sales_df['Year'] = sales_df['InvoiceDate'].dt.year
sales_df['Month'] = sales_df['InvoiceDate'].dt.month
sales_df['MonthName'] = sales_df['InvoiceDate'].dt.strftime('%b')
sales_df['YearMonth'] = sales_df['InvoiceDate'].dt.to_period('M').astype(str)

# Check the new fields
display(
    sales_df[
        ['InvoiceDate', 'Year', 'Month', 'MonthName', 'YearMonth']
    ].head(10)
)

print("\nYears:", sorted(sales_df['Year'].unique()))
print("Number of months:", sales_df['YearMonth'].nunique())
```

```python
# Monthly revenue analysis

monthly_sales = (
    sales_df.groupby('YearMonth')['Revenue']
    .sum()
    .reset_index()
)

display(monthly_sales)
```

```python
# Check the date range in each month

monthly_dates = (
    sales_df.groupby('YearMonth')['InvoiceDate']
    .agg(['min', 'max', 'count'])
    .reset_index()
)

display(monthly_dates)
```

```python
# Overall date range of the analysis dataset

print("First transaction:", sales_df['InvoiceDate'].min())
print("Last transaction:", sales_df['InvoiceDate'].max())
print("Total days:",
      (sales_df['InvoiceDate'].max() - sales_df['InvoiceDate'].min()).days)
```

```python
# Top 10 products by net revenue

top_products = (
    sales_df.groupby(['StockCode', 'Description'])['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

display(top_products)
```

```python
# Revenue by country

country_revenue = (
    sales_df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

display(country_revenue)
```

```python
# Top 10 customers by net revenue

top_customers = (
    sales_df.dropna(subset=['Customer ID'])
    .groupby('Customer ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

display(top_customers)
```

```python
# Customer revenue concentration

customer_revenue = (
    sales_df.dropna(subset=['Customer ID'])
    .groupby('Customer ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
)

top_10_revenue = customer_revenue.head(10).sum()
total_customer_revenue = customer_revenue.sum()
top_10_percentage = (top_10_revenue / total_customer_revenue) * 100

print("Total customer-attributed revenue: £{:,.2f}".format(total_customer_revenue))
print("Top 10 customers revenue: £{:,.2f}".format(top_10_revenue))
print("Top 10 customers revenue share: {:.2f}%".format(top_10_percentage))
```

```python
# Analyze cancellations / returns

returns = sales_df[sales_df['Quantity'] < 0].copy()
sales = sales_df[sales_df['Quantity'] > 0].copy()

return_value = returns['Revenue'].sum()
gross_sales_value = sales['Revenue'].sum()
net_revenue = sales_df['Revenue'].sum()

return_rate = len(returns) / len(sales_df) * 100
return_value_pct = abs(return_value) / gross_sales_value * 100

print("Positive sales revenue: £{:,.2f}".format(gross_sales_value))
print("Return/cancellation value: £{:,.2f}".format(return_value))
print("Net revenue: £{:,.2f}".format(net_revenue))
print("Return transaction rate: {:.2f}%".format(return_rate))
print("Return value as % of positive sales: {:.2f}%".format(return_value_pct))
```

```python
# Top 10 products by return/cancellation value

product_returns = (
    returns.groupby(['StockCode', 'Description'])['Revenue']
    .sum()
    .sort_values()
    .head(10)
    .reset_index()
)

display(product_returns)
```

```python
# Identify non-product adjustment transactions

adjustment_codes = ['M', 'AMAZONFEE', 'BANK CHARGES', 'D', 'POST', 'DOT']

adjustments = returns[returns['StockCode'].isin(adjustment_codes)].copy()
product_returns = returns[~returns['StockCode'].isin(adjustment_codes)].copy()

print("Operational/financial adjustment value: £{:,.2f}".format(
    adjustments['Revenue'].sum()
))

print("Product-related return/cancellation value: £{:,.2f}".format(
    product_returns['Revenue'].sum()
))

print("Operational/financial adjustment rows:", len(adjustments))
print("Product-related return/cancellation rows:", len(product_returns))
```

```python
# Top 10 products by product-related return/cancellation value

top_product_returns = (
    product_returns.groupby(['StockCode', 'Description'])['Revenue']
    .sum()
    .sort_values()
    .head(10)
    .reset_index()
)

display(top_product_returns)
```

```python
# Return rate for the top-selling products

product_sales = (
    sales[sales['StockCode'].isin(top_products['StockCode'])]
    .groupby(['StockCode', 'Description'])['Revenue']
    .sum()
)

product_return_value = (
    product_returns[product_returns['StockCode'].isin(top_products['StockCode'])]
    .groupby(['StockCode', 'Description'])['Revenue']
    .sum()
)

return_rate_top_products = pd.concat(
    [product_sales, product_return_value],
    axis=1
)

return_rate_top_products.columns = ['SalesRevenue', 'ReturnValue']
return_rate_top_products = return_rate_top_products.fillna(0)

return_rate_top_products['ReturnRate'] = (
    abs(return_rate_top_products['ReturnValue'])
    / return_rate_top_products['SalesRevenue']
    * 100
)

return_rate_top_products = (
    return_rate_top_products
    .sort_values('ReturnRate', ascending=False)
    .reset_index()
)

display(return_rate_top_products)
```

```python
#  Create analysis-ready processed dataset

df_processed = df.copy()

# 1. Remove exact duplicate rows
df_processed = df_processed.drop_duplicates().copy()

# 2. Create transaction value
df_processed['TotalAmount'] = (
    df_processed['Quantity'] * df_processed['Price']
)

# 3. Remove rows with missing description and zero price
df_processed = df_processed[
    ~(
        df_processed['Description'].isna() &
        (df_processed['Price'] == 0)
    )
].copy()

# 4. Remove negative-price accounting adjustments from sales analysis
#    (These are non-sales financial adjustments)
df_processed = df_processed[df_processed['Price'] >= 0].copy()

# 5. Create time-based fields
df_processed['Year'] = df_processed['InvoiceDate'].dt.year
df_processed['Month'] = df_processed['InvoiceDate'].dt.month
df_processed['MonthName'] = df_processed['InvoiceDate'].dt.strftime('%B')
df_processed['YearMonth'] = df_processed['InvoiceDate'].dt.to_period('M').astype(str)

print("Processed rows:", len(df_processed))
print("Processed columns:", len(df_processed.columns))
print("Remaining missing descriptions:", df_processed['Description'].isna().sum())
print("Remaining negative prices:", (df_processed['Price'] < 0).sum())
```

```python
#  Create sales analysis dataset

sales_df = df_processed[
    (df_processed['Price'] > 0) &
    (df_processed['Quantity'] != 0)
].copy()

# Revenue
sales_df['Revenue'] = (
    sales_df['Quantity'] * sales_df['Price']
)

print("Sales analysis rows:", len(sales_df))
print("Total net revenue: £{:,.2f}".format(sales_df['Revenue'].sum()))
print("Positive sales revenue: £{:,.2f}".format(
    sales_df.loc[sales_df['Quantity'] > 0, 'Revenue'].sum()
))
print("Negative transaction value: £{:,.2f}".format(
    sales_df.loc[sales_df['Quantity'] < 0, 'Revenue'].sum()
))
```

```python
import os

# Create output folder
os.makedirs('/content/assessment_outputs', exist_ok=True)

# 1. Monthly revenue
monthly_revenue = (
    sales_df.groupby('YearMonth')['Revenue']
    .sum()
    .reset_index()
)

# 2. Country revenue
country_revenue = (
    sales_df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

# 3. Top products
top_products_export = (
    sales_df.groupby(['StockCode', 'Description'])['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(20)
    .reset_index()
)

# 4. Top customers
top_customers_export = (
    sales_df.dropna(subset=['Customer ID'])
    .groupby('Customer ID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(20)
    .reset_index()
)

# 5. Product return/cancellation impact
product_returns_export = (
    product_returns
    .groupby(['StockCode', 'Description'])['Revenue']
    .sum()
    .sort_values()
    .head(20)
    .reset_index()
)

# 6. KPI summary
kpi_summary = pd.DataFrame({
    'Metric': [
        'Original rows',
        'Processed rows',
        'Sales analysis rows',
        'Net revenue',
        'Positive sales revenue',
        'Negative transaction value',
        'Customer-attributed revenue',
        'Top 10 customer revenue share'
    ],
    'Value': [
        len(df),
        len(df_processed),
        len(sales_df),
        sales_df['Revenue'].sum(),
        sales_df.loc[sales_df['Quantity'] > 0, 'Revenue'].sum(),
        sales_df.loc[sales_df['Quantity'] < 0, 'Revenue'].sum(),
        customer_revenue.sum(),
        top_10_percentage
    ]
})

# Export files
monthly_revenue.to_csv(
    '/content/assessment_outputs/monthly_revenue.csv', index=False
)

country_revenue.to_csv(
    '/content/assessment_outputs/country_revenue.csv', index=False
)

top_products_export.to_csv(
    '/content/assessment_outputs/top_products.csv', index=False
)

top_customers_export.to_csv(
    '/content/assessment_outputs/top_customers.csv', index=False
)

product_returns_export.to_csv(
    '/content/assessment_outputs/product_returns.csv', index=False
)

kpi_summary.to_csv(
    '/content/assessment_outputs/kpi_summary.csv', index=False
)

print("All analysis files exported successfully.")
print("\nFiles created:")
for file in os.listdir('/content/assessment_outputs'):
    print("-", file)
```
